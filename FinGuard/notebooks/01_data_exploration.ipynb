## Notebook 01: Exploratory Data Analysis & Feature Distributions

**Objective**: Survey dataset distributions, examine monetary volume disparities between legitimate and fraudulent transactions, evaluate feature correlations, and quantify business financial exposure.

```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

ROOT = Path("..").resolve()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.exploratory_analysis import (
    load_clean_data,
    compute_eda_summary_stats,
    plot_class_distribution,
    plot_amount_distribution,
    plot_correlation_matrix,
)

data_path = ROOT / "data" / "processed" / "clean_creditcard.csv"
fig_dir = ROOT / "reports" / "figures"
df = load_clean_data(data_path)
print(f"Loaded clean transactions: {len(df):,} rows x {len(df.columns)} columns")
```

### 1. Statistical Volume & Turnover Summary

```python
stats = compute_eda_summary_stats(df)
vol = stats['volume']
exp = stats['financial_exposure']
amt = stats['amount_metrics']

print("=== TRANSACTION VOLUME ===")
print(f"Total: {vol['total_transactions']:,}")
print(f"Legitimate: {vol['legitimate_count']:,} | Fraud: {vol['fraud_count']:,}")
print(f"Fraud Incidence Rate: {vol['fraud_percentage']:.4f}% ({vol['imbalance_ratio']})")

print("\n=== FINANCIAL EXPOSURE ===")
print(f"Total Turnover: €{exp['total_turnover_eur']:,.2f}")
print(f"Fraud Turnover: €{exp['fraud_exposure_eur']:,.2f} ({exp['fraud_exposure_percentage']:.4f}%)")

print("\n=== TRANSACTION AMOUNT METRICS ===")
print(f"Legitimate Mean: €{amt['legit_mean']:.2f} (Median: €{amt['legit_median']:.2f}, Max: €{amt['legit_max']:,.2f})")
print(f"Fraudulent Mean: €{amt['fraud_mean']:.2f} (Median: €{amt['fraud_median']:.2f}, Max: €{amt['fraud_max']:,.2f})")
```
